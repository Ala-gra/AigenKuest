Build the transition dictionary

```python
transitions = {}

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]

    if current_word not in transitions:
                    transitions[current_word] = {}

    if next_word not in transitions[current_word]:
                                transitions[current_word][next_word] = 0

    transitions[current_word][next_word] += 1

print(transitions)
```

Calculate  the transition probabilities

```python
law_transitions = transitions.get("law", {})

total = sum(law_transitions.values())

for word, count in law_transitions.items():
    probability = count / total
    print(word, round(probability, 3))
```

Use the probabilities to generate text from your book.

```python
import random

def generate_text(start_word, length):
    current_word = start_word
    result = [current_word]

    for _ in range(length - 1):
        next_words = transitions.get(current_word, {})
        if not next_words:
            break
        words_list = list(next_words.keys())
        counts = list(next_words.values())
        current_word = random.choices(words_list, weights=counts, k=1)[0]
        result.append(current_word)

    return " ".join(result)
```

**IMPROVING THE MODEL TO CONSIDER TWO CONSECUTIVE WORDS**

Our current model considers only one previous word. We'll upgrade it to consider two consecutive words, which can help it produce more coherent text.

1. Let's build our two-word transition dictionary.

```python
transitions_2 = {}

for i in range(len(words) - 2):
    current_pair = (words[i], words[i + 1])
    next_word = words[i + 2]

    if current_pair not in transitions_2:
        transitions_2[current_pair] = {}

    if next_word not in transitions_2[current_pair]:
        transitions_2[current_pair][next_word] = 0

    transitions_2[current_pair][next_word] += 1
```

```python
print(transitions_2.get(("the", "law"), {}))
```

2. Let's calculate the probabilities


```python
pair = ("the", "law")
next_words = transitions_2.get(pair, {})

total = sum(next_words.values())

for word, count in next_words.items():
    print(word, count / total)
```

3. Build the second-order generator


```python
import random

def generate_text_2(start_words, length):
    current_pair = tuple(start_words.split())
    result = list(current_pair)

    for _ in range(length - 2):
        next_words = transitions_2.get(current_pair, {})

        if not next_words:
            break

        words_list = list(next_words.keys())
        counts = list(next_words.values())

        next_word = random.choices(
            words_list,
            weights=counts,
            k=1
        )[0]

        result.append(next_word)
        current_pair = (current_pair[1], next_word)

    return " ".join(result)
```

5.Handling dead ends

```python
import random

def generate_text_2(start_words, length):
    current_pair = tuple(start_words.split())
    result = list(current_pair)

    # Store all pairs that have possible next words
    valid_pairs = list(transitions_2.keys())

    for _ in range(length - 2):
        next_words = transitions_2.get(current_pair, {})

        if not next_words:
            # Restart using a randomly selected pair
            current_pair = random.choice(valid_pairs)
            result.extend(current_pair)
            continue

        words_list = list(next_words.keys())
        counts = list(next_words.values())

        next_word = random.choices(
            words_list,
            weights=counts,
            k=1
        )[0]

        result.append(next_word)
        current_pair = (current_pair[1], next_word)

    return " ".join(result)
```

```python
import numpy as np

states = list(transitions.keys())[:5]

P = np.zeros((5, 5))

for i, current_word in enumerate(states):
    next_words = transitions[current_word]
    total = sum(next_words.values())

    for j, next_word in enumerate(states):
        if next_word in next_words:
            P[i, j] = next_words[next_word] / total

print("States:", states)
print(P)
```

```python
import numpy as np
from scipy.sparse import lil_matrix

states = list(transitions.keys())
word_to_index = {word: i for i, word in enumerate(states)}

P = lil_matrix((len(states), len(states)), dtype=float)

for current_word, next_words in transitions.items():
    i = word_to_index[current_word]
    total = sum(next_words.values())

    for next_word, count in next_words.items():
        if next_word in word_to_index:
            j = word_to_index[next_word]
            P[i, j] = count / total

P = P.tocsr()

print("Matrix shape:", P.shape)
print("Nonzero entries:", P.nnz)
print("First five row sums:", np.asarray(P.sum(axis=1)).flatten()[:5])
```

```python
from scipy.sparse.linalg import eigs

# Find the eigenvector associated with eigenvalue 1
eigenvalues, eigenvectors = eigs(P.T, k=1, which="LM")

# Extract the eigenvector
steady_state = np.real(eigenvectors[:, 0])

# Normalize so probabilities sum to 1
steady_state = steady_state / steady_state.sum()

print("Sum of probabilities:", steady_state.sum())

# Find the five words with the largest steady-state probabilities
top_indices = np.argsort(steady_state)[-5:][::-1]

for i in top_indices:
    print(states[i], ":", steady_state[i])
```

```python
check = steady_state @ P

print("Maximum difference:", np.max(np.abs(check - steady_state)))
```

```python
# Calculate actual word frequencies
total_words = len(words)

for i in np.argsort(steady_state)[-5:][::-1]:
    word = states[i]
    frequency = words.count(word) / total_words
    print(
        word,
        "| Actual frequency:", round(frequency, 4),
        "| Steady state:", round(steady_state[i], 4)
    )
```
